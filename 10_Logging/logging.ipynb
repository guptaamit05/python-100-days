```python
## Configure basic logging..

import datetime

dt = datetime.datetime.now()
print(dt)
logging.basicConfig(level=logging.DEBUG)

logging.debug(f"{dt} this is debug message...")
logging.info( f"{dt} this in info msg")
logging.warning(f"{dt} this is warning msg")
logging.error( f"{dt} this is error msg")
logging.critical( f"{dt} this is critical msg")
```
