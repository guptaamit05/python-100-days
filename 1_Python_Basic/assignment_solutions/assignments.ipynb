```python
#12) Write a program that checks if a number input by the user is a prime number using a for loop.
user_input = int(input("enter any number"))
print(user_input)
is_prime = True
if user_input <= 1 :
    is_prime = False
else:
    for i in range(2, int(user_input**0.5)+1):
        if(user_input%i==0):
            is_prime = False
            break
print("Prime") if is_prime else print("Not Prime")
```
