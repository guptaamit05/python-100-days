```python
#### Regular Expression syntax keywords..

# \d  only numbers
# \w  only alphanumerice.
# \s  only white spaces
# \S  non spaces
# \D  non digits
# \W  non alphanumerice

#  *      for zero on more times
#  +      for one or more times
#  {3}    for exact 3 times
#  {2,4}  occurs 2 or 4 times 
#   ?     onces or none
# {3,}    Occurs 3 or more time

#  ^      string starts with
#  $      for ends with
```

```python
url = "https://quotes.toscrape.com/page/{}/"
all_top_tags = set()
srcc = requests.get(url.format(1))
soup = bs4.BeautifulSoup(srcc.text, "lxml")
author_names = set()
quote_list=[]

[quote_list.append(quote.contents[1].text) for quote in soup.select(".quote")]
{author_names.add(author.contents[0]) for author in soup.select('.author')}
[all_top_tags.add(item.contents[1].text) for item in soup.select(".tag-item")]

print(quote_list)    
print(author_names)
print(all_top_tags)
```

```python
unique_authors = set()
page = 1
while(True):
    url_all = "https://quotes.toscrape.com/page/{}/"
    srcc = requests.get(url_all.format(page))
    soup_all = bs4.BeautifulSoup(srcc.text, 'lxml')
    is_author_class = soup_all.select('.author')
    print(len(is_author_class))
    if(len(is_author_class) == 0):
        break
    {unique_authors.add(author.contents[0]) for author in soup_all.select('.author')}
    page+=1

print(unique_authors)
```
