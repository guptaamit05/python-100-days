```python
class Bank:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        if amount>0:
            self.balance+=amount
    def withdraw(self, amount):
        if self.balance> amount and amount > 0:
            self.balance -=amount
    
    def __str__(self):
        return f"Current balance: {self.balance}"
        
obj = Bank("Dhiraj", 1000)

print(obj)
obj.withdraw(200)
obj.withdraw(200)
obj.withdraw(200)
print(obj)
obj.deposit(3000)
print(obj)
```
