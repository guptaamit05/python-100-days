```python
# Create a custom iterator class named `Countdown` that takes a number and counts down to zero. Implement the `__iter__` and `__next__` methods. Test the iterator by using it in a for loop.

class Countdown:
    def __init__(self, num):
        self.num = num
    
    def __iter__(self):
        return self
    def __next__(self):
        if self.num<= 0:
            raise StopIteration
        else:
            self.num -=1
            return self.num

for n in Countdown(10):
    print(n)
```

```python
# Write a generator function named `fibonacci` that yields the Fibonacci sequence. Test the generator by iterating over it and printing the first 10 Fibonacci numbers.

def custom_generator(n):
    a,b = 0,1
    for _ in range(n):
        yield a
        a,b = b, a+b
    
for x in custom_generator(5):
    print(x)
```

```python
# Write two generator functions: `even_numbers` that yields even numbers up to a limit, and `squares` that yields the square of each number from another generator. Chain these generators to produce the squares of even numbers up to 20.

def even_numbers(limt):
    for n in range(limt+1):
        if n%2==0:
            yield n
def square(num):
    for n in num:
        yield n*n

evn = even_numbers(20)
sq = square(evn)
for sqq in sq:
    print(sqq)
```

```python
# Write a decorator named `time_it` that measures the execution time of a function. Apply this decorator to a function that calculates the factorial of a number.
import datetime

def time_it(fun):
    def wrapper(*args, **kwargs):
        start = datetime.datetime.now()
        result = fun(*args)
        end = datetime.datetime.now()
        return result, (end-start)
    return wrapper

def factorial(num):
    if num <=0:
        return 1
    return (num * factorial(num-1))

@time_it
def get_facto(n):
    return factorial(n)

res, timediff = get_facto(12)
print(res, timediff)
```

```python
# Write a decorator named `repeat` that takes an argument `n` and repeats the execution of the decorated function `n` times. Apply this decorator to a function that prints a message.

def repeat(num):
    def decorator_repeat(fun):
        def wrapper(*args, **kwargs):
            for _ in range(num):
                fun(*args, **kwargs)
        return wrapper
    return decorator_repeat

@repeat(5)
def myDispay(msg):
    print(msg)

(myDispay("Okay Amar khar"))
```

```python
# Write two decorators: `uppercase` that converts the result of a function to uppercase, and `exclaim` that adds an exclamation mark to the result of a function. Apply both decorators to a function that returns a greeting message.

def uppercase(fun):
    def wrapper(*args):
        res = fun(*args)
        return res.upper()
    return wrapper

def exclaim(fun):
    def wrapper(*args):
        res =  fun(*args)
        return res +"!"
    return wrapper

input_user = input("enter something")
@uppercase
@exclaim
def greeting(input_user):
    return input_user

greeting(input_user)
```

```python
# Create a class decorator named `singleton` that ensures a class has only one instance. Apply this decorator to a class named `DatabaseConnection` and test it.

def singleton(cls):
    instance = {}
    def get_instance(*args, **kwargs):
        if cls not in instance:
            instance[cls] = cls(*args, **kwargs)
        return instance[cls]
    return get_instance

@singleton
class DatabaseConnection:
    def __init__(self):
        print("DB connection is created.")
        
ob = DatabaseConnection()
ob3 = DatabaseConnection()
print(ob is ob3)
    
```

```python
# Create a custom iterator class named `ReverseString` that iterates over a string in reverse. Write a decorator named `uppercase` that converts the string to uppercase before reversing it. Apply the decorator to the `ReverseString` class.


def upperCase(cls):
    
    class Wrapper(cls):
        def __init__(self, *args, **kwargs):
            super().__init__(*args, **kwargs)
            self.inpt = self.inpt.upper()
            print(self.inpt)
    return Wrapper
    
@upperCase
class ReverseString:
    def __init__(self, inpt):
        self.inpt = inpt
        self.index = len(inpt)
        
    def __iter__(self):
        return self
    def __next__(self):
        if self.index ==0:
            raise StopIteration
        self.index-=1
        return self.inpt[self.index]

ob = ReverseString("hello amit")
strr = ""
for char in ob:
    strr +=char
    
print(strr)
```

```python
# Write a stateful generator function named `counter` that takes a start value and increments it by 1 each time it is called. Test the generator by iterating over it and printing the first 10 values.
def counter(start_value):
    current = start_value
    while True:
        yield current
        current+=1

count = counter(0)
for _ in range(10):
    print(next(count))
```

```python
# Write a generator function named `safe_divide` that takes a list of numbers and yields the division of each number by a given divisor. Implement exception handling within the generator to handle division by zero.

def sae_divide(num, divisor):
    for x in num:
        try:
            yield x/divisor
        except ZeroDivisionError as e:
            yield f" {e}"

for resu in sae_divide([1,2,3,4,5], 0):
    print(resu)   
```

```python
#Write a decorator named `open_file` that manages the opening and closing of a file. Apply this decorator to a function that writes some text to a file.

def open_file(file_name, mode):
    def decorator(fun):
        def wrapper(*args, **kwargs):
            with open(file_name, mode) as f:
                return fun(f, *args, **kwargs)
        return wrapper
    return decorator
            
@open_file('sample.txt', "w")
def write_to_file(f, txt_to_write):
    try:
        f.write(txt_to_write)
    except Exception as e :
        print(f"Error: {e}")
        
write_to_file("Hi amit how are you..................?")
    
```

```python
# Create an infinite iterator class named `InfiniteCounter` that starts from a given number and increments by 1 indefinitely. Test the iterator by printing the first 10 values generated by it.

class InfiniteCounter:
    def __init__(self, num):
        self.count = num
    
    def __iter__(self):
        return self
    def __next__(self):
        self.count +=1
        return self.count

counter = InfiniteCounter(0)
for _ in range(10):
    print(next(counter))


```

```python
# Write three generator functions: `integers` that yields integers from 1 to 10, `doubles` that yields each integer doubled, and `negatives` that yields the negative of each doubled value. Chain these generators to create a pipeline that produces the negative doubled values of integers from 1 to 10.

def integers(num):
    for n in range(1, num+1):
        yield(n)

def doubles(num):
    yield num*2

def negative(num):
    yield (-num)

for g in integers(10):
    nu = next(negative(next(doubles(g))))
    print(nu)
    # res = next(negative(nu))
    # print(res)
```
