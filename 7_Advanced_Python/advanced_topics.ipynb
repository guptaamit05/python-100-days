```python
## Decorators...

def deco(n):
    def decorat(fun):
        def wrapper(name):
            for _ in range(n):
                fun(name)
        return wrapper
    return decorat
            
@deco(5)
def my_name(name):
    print(f'my name is {name}')

my_name("Satish")
```

```python

def upperrr(fun):
    def wrapper():
        v = fun().upper().split(" ")
        return "__".join(v)
    return wrapper

def expo(fun):
    def wrapper():
        return fun()+"!!!"
    return wrapper

@upperrr
@expo
def greet():
    return "hello goa full on.."


print(greet())
```

```python
## static method...

class myMathUtils:
    @staticmethod
    def mystaticmethod(a,b):
        return a+b

# using class name calling static method..
t = myMathUtils.mystaticmethod(55,44)
print(t)

#using object calling static method...
ob = myMathUtils()
le = ob.mystaticmethod(10,30)
print(le)
```
