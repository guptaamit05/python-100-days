```python
lst = [1,2,3,4,5,6]
lst
```

```python
def iterate_lst(lst):
    for lp in lst:
        yield(lp)
```

```python
gen_list = iterate_lst(lst)
# next(gen_list)
```

```python
for value in iterate_lst(lst):
    print(value)
print("................")
for value in iterate_lst(lst):
    print(value)
```

```python
# Function Copy..
def one_fun(one=90):
    print("Hi from one fun ", one)

two = one_fun   # Function copy...
one_fun(100)
two(45)
del one_fun
two(33)
```

```python
# Closure example
def outer_fun():
    name = "Harish Das"
    def inner_fun():
        print("Calling from inner fun:---",name)
    return inner_fun

calling_outer = outer_fun()   
calling_outer()
```

```python
##  Decorator fucction...

def decorator_user(user_fun):
    def wrapper():
        print("I am a decorator function.. BEFORE.")        
        user_fun()
        print("I am a decorator function.. AFTER....")        
        
    return wrapper
        

@decorator_user
def user_fun():
    print("Hi Amit first call decorator...")

user_fun()
    
```
