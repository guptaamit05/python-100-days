```python
## Write a recursive function  fibonacci(n)  that prints the first  n  Fibonacci numbers

def fibonancci(n):
    if(n <= 1):
        return n
    return fibonancci(n-1) + fibonancci(n-2)

n = 10
for i in range(n):
    print(fibonancci(i))
```
