```python
def even_or_odd(n):   # funtion with param..
    """ 
    This is a function that check number is even or odd..
    """
    return "Even" if n%2==0 else "Odd"
even_or_odd(2)
# evn_odd(7)
```

```python
## Variable length Arguments.
# Positional Arguments.
def sum(*args):  ## we can take any name instead of args.. 
    for n in args:
        print(f"number = {n}")
print(sum(3,4,5))
```

```python
## Keywords arguments..   single * is POsitional Start and ** is keywords arguments
def print_details(*args, **kwargs):
    for value in args:
        print(f"Value = {value}")
    for k,v in kwargs.items():
        print(k,v)
print(print_details(1,2,3,4, name="punit", age=34, city="Indore"))
```
