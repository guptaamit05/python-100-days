```python
def check_unique(lst):
    # Your code goes here
    is_unique = True
    unique_item = []
    for x in lst:
        if x not in unique_item:
            unique_item.append(x)
            continue
        is_unique = False
        break
    return is_unique

check_unique([1,2,3,4,5,5])
```

```python
def check_unique(lst):
    un = set()
    for x in lst:
        if x in un:
            return False
        un.add(x)
    return True
            
check_unique([1,2,3,4,5])
```

```python
def max_consecutive_difference(lst):
    if len(lst) < 2:
        return None

    max_diff = 0
    pair = None

    for i in range(len(lst) - 1):
        diff = abs(lst[i] - lst[i+1])
        if diff > max_diff:
            max_diff = diff
            pair = (lst[i], lst[i+1])

    return max_diff, pair

lee = max_consecutive_difference(lst)
print(lee[0])
print(lee[1])
```

```python
def merge_two_sorted_lists(list1, list2):
    i, j=0, 0
    result = []
    while  i<len(list1) and j<len(list2):
        if list1[i]<list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
            
    print(result)
    result.extend(list1[i:])
    result.extend(list2[j:])
    return result

list1 = [1, 4, 7]
list2 = [2, 3, 5, 8]
merge_two_sorted_lists(list1, list2)
```

```python
def rotate_list(lst, k):
    n = len(lst)
    k = k%n
    print(k)
    for _ in range(k):
        last = lst.pop()
        lst.insert(0, last)
    return lst

lst = [1, 2, 3, 4, 5]
k = 2
# lst = [10, 20, 30, 40, 50], k = 3
rotate_list(lst, k)
```

```python
def merge_lists_to_dictionary(keys, values):
    if len(keys) != len(values):
        return False
    result = {}
    for i in range(len(keys)):
        result[keys[i]] = values[i]

    return result

keys = ['a', 'b', 'c']
values = [1, 2, 3]
merge_lists_to_dictionary(keys, values)
# Output: {'a': 1, 'b': 2, 'c': 3}
```

```python
# Input: "hello world hello"
# Output: {'hello': 2, 'world': 1}
def count_word_frequency(sentence):
    # Your code goes here
    lst = sentence.split(" ")
    result= {}
    for x in lst:
        if result.get(x):
            result[x] +=1
        else:
            result[x] = 1
    return result
st = "hello world hello this is my first. this is my pen"
count_word_frequency(st)
```

```python
def is_palindromic_tuple(tup):
    # Your code goes here
    lst = list(tup)
    rev_lis = [x for x in lst[::-1]]
    
    if lst == rev_lis:
        return True
    else:
        return False

# Input = (1, 2, 3, 2, 1)
# Input = ('a', 'b', 'c', 'b', 'a')
# Input = (1, 2, 3, 4, 5)
Input = ('x', 'y', 'z', 'x')
# Input = ('a',)
is_palindromic_tuple(Input)
```
