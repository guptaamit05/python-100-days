```python
#1) Define a recursive function to calculate the nth Fibonacci number using memoization. Test the function with different inputs.

def testFibno(n, memo={}):
    if n in memo:
        return memo[n]
    if n<=1:
        return n
    memo[n] = testFibno(n-1) + testFibno(n-2)
    return memo[n]

testFibno(10)
```

```python
#3) Define a function that takes a variable number of keyword arguments and returns a dictionary containing only those key-value pairs where the value is an integer. Test the function with different inputs.

def kwargFun(**kwargs):
    return {k:v for k,v in kwargs.items() if isinstance(v, int)}

kwargFun(name='harish', age=33, test=1, om=44, car='bike')
```

```python
# 4) Define a function that takes another function as a callback and a list of integers. The function should apply the callback to each integer in the list and return a new list with the results. Test with different callback functions.

def first(n):
    return n*2

def second(first, int_list):
    return [first(x) for x in int_list if first(x) ]

second(lambda x: x%2==0, [1,2,3,4])
```

```python
# 6) Define a function that calculates the time taken to execute another function. Apply this decorator to a function that performs a complex calculation. Test the decorated function with different inputs.

import time

def decoratorFun(anotherFun):
    def wrapper():
        print("first run")
        st = time.time()
        anotherFun()
        
        print("second run")
        end =time.time()
        return end-st
        
    return wrapper

@decoratorFun
def displayName():
    time.sleep(1)
    print("hi harish how are you..")
    
    

displayName()
```

```python
#7) Define a higher-order function that takes two functions, a filter function and a map function, along with a list of integers. The higher-order function should first filter the integers using the filter function and then apply the map function to the filtered integers. Test with different filter and map functions.


def fitMap(filterfun, mapfun, lst):
    return [mapfun(x) for x in lst  if filterfun(x)]

def filterfun(n):
    return n%2==0

def mapfun(n):
    return n+2

fitMap(filterfun, mapfun, [1,2,3,4,5,6,7,8,9])
```

```python
#8) Define a function that composes two functions, f and g, such that the result is f(g(x)). Test with different functions f and g.

def composefun(f,g):
    return lambda x: f(g(x))

h = composefun(lambda x: x+1, lambda y: y+2)
h(5)
```

```python
#11) Define a function that generates an infinite sequence of Fibonacci numbers. Test by printing the first 10 numbers in the sequence.

def infinteFun(n):
    a,b =0,1
    while True:
        yield a
        a,b = b, a+b
t = infinteFun(10)
for _ in range(10):
    print(next(t))
```

```python
#12) Define a curried function that takes three arguments, one at a time, and returns their product. Test the function by providing arguments one at a time.

def curryFun(a):
    def inner(x):
        def innermost(y):
            return a*x*y
        return innermost
    return inner
        
x = curryFun(10)(20)(30)
print(x)
```

```python
def write_to_file(file_name, lst):
    try:
        with open(file_name, 'w') as f:
            for item in lst:
                f.write(f"Item: {item} \n")
    except Exception as e:
        print("Something went wrong: ",e)
        
write_to_file("sample.txt", [1,2,3,4,5])
```

```python
mixture = ['amit', 22.4, 2,33,55,66,'om', 'sai']

def filterData(lst):
    strr, floatValue, intValue = [], [],[]
    for item in lst:
        if isinstance(item, str):
            strr.append(item)
        if isinstance(item, float):
            floatValue.append(item)

        if isinstance(item, int):
            intValue.append(item)
    return strr, floatValue, intValue

filterData(mixture)
        
```
