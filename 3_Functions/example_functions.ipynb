```python
def strong_or_weak_pass(pw):
    if len(pw)<8:
        return False
    if not any(char.isdigit() for char in pw):
        return False
    if not any(char.islower() for char in pw):
        return False
    if not any(char.isupper() for char in pw):
        return False
    if not any(char in "!@#$%^&89)_+" for char in pw):
        return False
    return True

print(strong_or_weak_pass("9999889aA_8"))
```

```python
## Calculate total cost of item in shoping cart..
def sum_items(itm):
    total = 0
    for item in itm:
        total += item['qty'] * item['price']
    return total

items = [
    {"name": "apple", "price": 34.243534, "qty": 3},
    {"name": "banana", "price": 12.5, "qty": 6},
    {"name": "orange", "price": 25.0, "qty": 4},
    {"name": "mango", "price": 60.0, "qty": 2},
    {"name": "grapes", "price": 45.5, "qty": 5},
    {"name": "pineapple", "price": 80.0, "qty": 1}
]
sum = sum_items(items)
print(sum)

```

```python
## check palindrome...

def is_palindrom(strr):
    s = strr.lower().replace(" ", "")
    return s==s[::-1]

is_palindrom("namsan")
```

```python
## Factorial of a number..
def fact(n):
    if n==0:
        return 1
    else: 
        return n*fact(n-1)

fact(10)
```

```python
## VALIDATE EMAIL ADDRESS..
import re
def is_valid_email(email):
    rg = r"^[a-zA-Z0-9-+._]+@[a-zA-Z0-9-]+\.[a-zA-Z0-9-.]+$"  ## Regex for email check..
    is_match = re.match(rg, email)
    if is_match:
        return True
    return False

is_valid_email("amie@r.com")
```
