```python
#  count # of lines, # of words and # of char in a file...

pth = "example.txt"
def count_in_file(file_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()
        
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    print(line_count, word_count, char_count)
    return line_count, word_count, char_count


line_count, word_count, char_count = count_in_file(pth)
```

```python
class Bank:
    def __init__(self, initial_amount=0.00):
        self.initial_amount = initial_amount
    
    def log_transactions(self, transaction_string):
        with open("log_transaction.txt", "a") as f:
            f.write(f"\n {transaction_string} \n")
    
    def withdraw(self, amount):
        try:
            amount = float(amount)
            if(amount >self.initial_amount):
                self.log_transactions(f"insiffucient balance. current balance:{self.initial_amount} and user withdrawing: {amount} ")
                raise Exception("Insufficient Balance")
            else:
                self.initial_amount -= amount
        except Exception as e:
            print("Error: ", e)
    
    def deposit(self, amount):
        try:
            amount = float(amount)
        except ValueError:
            amount = 0
        self.initial_amount +=amount
        self.log_transactions(f"Bank Account me : {amount} add hua now current balance is : {self.initial_amount}")
    def printBalance(self):
        self.log_transactions("print method call kiya user ne..")
        return self.initial_amount

bank = Bank(50.20)

bank.deposit(10)
bank.withdraw(70)
print(bank.printBalance())
    
```
