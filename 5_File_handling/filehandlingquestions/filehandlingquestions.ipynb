```python
# Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

def countWords(fname):
    if(os.path.exists(fname)):
        with open(fname, 'r') as f:
            content = f.read().split()
            print(len(content))
    else:
        raise Exception("file not found..")
countWords("../sample.txt")
```

```python
def replaceWord(fname, oldtxt, newtxt):
    if(os.path.exists(fname)):
        with open(fname, 'r') as f:
            content = f.read()
        new_content = content.replace(oldtxt, newtxt)
        with open(fname, 'w') as f:
            f.write(new_content)
    else:
        raise Exception("file not found..")
replaceWord("../sample.txt", 'the', 'THE')
```

```python
def readReverse(fname):
    if(os.path.exists(fname)):
        with open(fname, 'r') as f:
            content = f.read().split("\n")
            content.reverse()
            new_cont = "\n".join(content)
            print(new_cont)
    else:
        raise Exception("file not found..")
readReverse("../sample.txt")
```

```python
def countEach(fname):
    if(os.path.exists(fname)):
        lines = []
        with open(fname, 'r') as f:
            lines = f.readlines()
            words = sum(len(ly.split()) for ly in lines)
            characters = sum(len(line) for line in lines)
            
            print (len(lines), (words), characters)
            
        # return f"Lines: {len(lines)}, Words = {len(words)}"
    else:
        raise Exception("file not found..")

countEach("../sample.txt")
```
