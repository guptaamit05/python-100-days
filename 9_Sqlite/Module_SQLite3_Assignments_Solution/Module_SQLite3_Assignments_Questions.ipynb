```python
# 1 Write a Python function to create a new SQLite3 database named `test.db`.
import sqlite3

def create_db():
    try:
        logger.info("creating db name test.db")
        conn = sqlite3.connect('test.db')
        logger.info("DB created successfuly..")
        conn.close()
    except Exception as e:
        print(f"Error: {e}")
        logger.error(f"database error: {e}")
    
logger.info("calling db creation funtion..")
create_db()

```

```python
## 2 Write a Python function to create a table named employees with columns id (integer), name (text), age (integer), and department (text) in the test.db database.

def create_table():
    try:
        logger.info("----------------------------------------------")
        logger.info("creating table..")
        db = sqlite3.connect("test.db")
        cursor = db.cursor()
        cursor.execute('create table employees(id integer primary key, name text, age integer, department text)')
        db.commit()
        db.close()
        logger.info("Table employees is created.")
    except Exception as e:
        logger.error(f"Error to create table: {e}")

create_table()
```

```python
## 3 Write a Python function to insert a new employee into the `employees` table.
## 4 Insert at least 5 different employees into the `employees` table.
def insert_data(data):
    try:
        logger.info(f"inserting data into table emmployees with data: {data}")
        con = sqlite3.connect('test.db')
        cursor = con.cursor()
        cursor.executemany("insert into employees(name, age, department) values(?, ? , ?)", data)
        con.commit()
        con.close()
        logger.info("inserted successfully..")
        print("Success")

    except Exception as e:
        logger.error(f"Failed to insert data: {e}")
        print("Failed to insert..")

insert_data([("Hari", 36, 'IT'), ('Punit', 49, 'Finance'), ('Mangal das', 36, 'CS')])
insert_data([("Nandu", 16, 'IT'), ('Rajesh patil', 29, 'Finance'), ('Ramdas', 32, 'CIVIL')])
```

```python
## 5 Write a Python function to fetch and display all records from the `employees` table.

conn = sqlite3.connect('test.db')
cursor = conn.cursor()

def get_emp_info():
    try:
        logger.info("getting all info of employees table")
        cursor.execute("select * from employees")
        rows = cursor.fetchall()
        logger.info(f"employees table has {len(rows)} rows")
        for row in (rows):
            print(f"Row-{row[0]} with  name: {row[1]}, having age:{row[2]} and in the department: {row[3]}")
        conn.commit()
        logger.info("employees data fetched successfully.")
    except Exception as e:
        logger.error(f"unable to fetch data: {e}")

get_emp_info()
```

```python
# 6 Write a Python function to fetch and display all employees from a specific department.

def fetch_data_by_dept(dept):
    try:
        logger.info(f"fetching data by department: {dept}")
        cursor.execute("select * from employees where department=?",(dept,))
        rows = cursor.fetchall()
        for row in rows:
            print(f"Department data: {row}")
        logger.info(f"data fetched successfully total rows: {len(rows)}")
    except Exception as e:
        print(f"{e}")
        logger.error(f"unable to fetch data : {e}")

fetch_data_by_dept('CS')
```

```python
# 10  Write a Python function to fetch and display employees older than a certain age.

def fetch_data_by_age(age):
    try:
        logger.info(f"fetching data by Age: {age}")
        cursor.execute("select * from employees where age > ?",(age,))
        rows = cursor.fetchall()
        for row in rows:
            print(f"Fething data by Age: {row}")
        logger.info(f"data fetched successfully total rows: {len(rows)}")
        print('data fetched successfully!')
    except Exception as e:
        print(f"{e}")
        logger.error(f"unable to fetch data : {e}")

fetch_data_by_age(29)
```

```python
# 11 Write a Python function to fetch and display employees whose names start with a specific letter.
def fetch_data_by_name_like(name):
    try:
        logger.info(f"fetching data by Name: {name}")
        cursor.execute('SELECT * FROM employees WHERE name LIKE ?', ('%' + name + '%',))
        rows = cursor.fetchall()
        for row in rows:
            print(f"Fething data by name: {row}")
        logger.info(f"data fetched successfully total rows: {len(rows)}")
        print(f'data fetched successfully! total rows:{len(rows)}')
    except Exception as e:
        print(f"{e}")
        logger.error(f"unable to fetch data : {e}")

fetch_data_by_name_like('das')
```

```python
import time
def fetch_employees_name_starts_with_performance(letter):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    start_time = time.time()
    cursor.execute('SELECT * FROM employees WHERE name LIKE ?', (letter + '%',))
    records = cursor.fetchall()
    end_time = time.time()
    conn.close()
    print("Time taken: {} seconds".format(end_time - start_time))
    for record in records:
        print(record)

# Test the function with the index
fetch_employees_name_starts_with_performance('A')
```
