```python
import sqlite3
```

```python
connection = sqlite3.connect("example.db")
connection
```

```python
## create cursor...

cursor = connection.cursor()
```

```python
cursor.execute(''' create table if not exists employees(
    id integer primary key,
    name text not null,
    age integer,
    department text    
    )''')
connection.commit()
```

```python
cursor.execute("select * from employees")
```

```python
cursor.execute("insert into employees(name, age, department) values('krish', 34, 'finance')")
cursor.execute("insert into employees(name, age, department) values('narayan', 39, 'HR')")
cursor.execute("insert into employees(name, age, department) values('gopal', 24, 'IT')")
cursor.execute("insert into employees(name, age, department) values('ram', 54, 'finance')")
connection.commit()
```

```python
cursor.execute("select * from employees")
```

```python
### Query...
cursor.execute('select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Update data
cursor.execute("update employees set age=34 where name='gopal'")
```

```python
## Delete data...

cursor.execute('delete from employees where id = 4')
cursor.execute('select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
#### bulk insert data....

cursor.execute('''create table if not exists product(
    id integer primary key,
    name text not null, 
    price integer)
''')
connection.commit()
```

```python
data = [('mobile', 2300), ('cooler',5600),('washing machine',20000),    ('fan', 2900)   ]
cursor.executemany(''' insert into product (name, price) values(?, ?) ''', data)
connection.commit()
```

```python
cursor.execute("select * from product")
rows = cursor.fetchall()

for row in rows:
    print(row)
```
