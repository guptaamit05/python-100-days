```python
## inheritance...

class Car:
    
    def __init__(self, window_num, engine_type):
        self.window_num = window_num
        self.engine_type = engine_type
    
    def drive(self):
        return f"Hi I am driving {self.window_num} windows car and it is {self.engine_type} engine type car.."
    
car = Car(4, "4 Stroke Engine")
car.drive()
```

```python
class TeslaCar(Car):
    def __init__(self, window_num, engine_type, is_self_drive):
        super().__init__(window_num, engine_type)
        self.is_self_drive = is_self_drive
    
    def selfDrive(self):
        return f"Hi from Tesla Car and it has {self.is_self_drive and 'Self Drive'} Feature "
    
tesla = TeslaCar(4,'electric', True)
tesla.selfDrive()
tesla.drive()
```

```python
## Enapsulation and Abstraction

#  Encapsulation  come with getter and setter methods 

### Public, Private and Protected variables...
class Person:
    def __init__(self, name , age):
        self.name = name # public variable
        self.age = age  # public variable
        
        

def get_person(person):
    return person.name

person = Person("amit", 34)

# person.name
get_person(person)   # amit
# dir(person)
```

```python
###  private variable

class Person:
    def __init__(self, name , age, city="Indore"):
        self.__name = name # private variable
        self.__age = age  # private variable
        self.city = city  # public variable
        
    

def get_person(person):
    return person.__name

person = Person("amit", 34)
# dir(person)

# person.name
# get_person(person)   # will get error.. AttributeError: 'Person' object has no attribute '__name'
```

```python
###  protected variable  ( it can be access in drived class but not direct accessible..)

class Person:
    def __init__(self, name , age, city="Indore"):
        self._name = name # protected variable
        self._age = age  # protected variable
        self.city = city  # protected variable
        
def get_person(person):
    return person._name

person = Person("amit", 34)
# dir(person)

get_person(person)   # will get error.. AttributeError: 'Person' object has no attribute '_name'
```

```python
## Encapsulation with getter and setter methods

class Person:
    def __init__(self, name , age):
        self.__name = name # private variable
        self.__age = age  # private variable
        
    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

person = Person("amit", 34)
print(person.get_name() )
person.set_name("Harish Das")
person.get_name() 
# dir(person)
```

```python
class Mycls:
    def __init__(self, name, email="test@test.com"):
        self.name=name
        self.email = email
    def showData(self):
        return (f"Hi, {self.name} okay {self.email} is your personal email?")
    
    def common(self):
        print("Hi from Parent...")

ob  = Mycls("Satish", "satish@gg.com")
ob.showData()

class childCls(Mycls):
    def __init__(self, name, age):
        super().__init__(name)
        self.age= age
    
    def getAge(self):
        return (f"This is my age: {self.age}")
    def common(self):
        print("Hi from Child Class...")
        
ob3 = childCls("narayan", 33)

ob3.getAge()
ob3.common()
ob.common()
```

```python
class Operator:
    def __init__(self, x, y=10):
        self.x = x
        self.y = y
    
    def __mul__(self, other):
        return (self.x * self.y * other)

b1 = Operator(2,3)
b3 = b1 * 12
b3
```

```python
class A:
    def __init__(self, name):
        self.name = name
    def start_db(self):
        print(f"Start the {self.name} Car.")

class B(A):
    def __init__(self, name):
        super().__init__(name)
        self.name = name
    def bike_start(self):
        print(f"{self.name} started..")
        
obb = B("Honda CB Shine")

obb.bike_start()
obb.start_db()
```
