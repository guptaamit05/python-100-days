```python
#9) Create a class named `MathOperations` with a static method to calculate the square root of a number. Call the static method without creating an object.
import math

class MathOperations:
    @staticmethod
    def square(num):
        return round(math.sqrt(num),3)
MathOperations.square(64)
```

```python
# 10) Create a class named `Rectangle` with private attributes `length` and `width`. Use properties to get and set these attributes. Create an object of the class and test the properties.

class Rectangle:
    def __init__(self, length, width):
        self.__length  = length
        self.__width = width
    
    @property
    def get(self):
        return self.__width, self.__length
    
    @get.setter
    def get(self, value):
        lent, wth = value
        self.__length = lent
        self.__width = wth
rec = Rectangle(20,10)
print(rec.get)
rec.get = (40,50)
print(rec.get)
```

```python
#11 Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

from abc import abstractmethod, ABC

class Shape(ABC):
    @abstractmethod
    def area(self):    
        pass

class Circle(Shape):
    def __init__(self, rad):
        self.rad = rad
        
    def area(self):
        return round(math.pi * self.rad, 4)
class Square(Shape):
    def __init__(self, edge):
        self.edge = edge
    def area(self):
        return round(self.edge**2)
    
circle = Circle(10)
print(circle.area())

square = Square(50)
print(square.area())
```
