```python
## Tuples methods..

# only 2 methods :  count, index
num = (2,3,4,5,6,7,2,3,4,5,6,2,3,4,5,6)
two_count = num.count(2)
print(two_count)

print(num.index(4))   ## 4 is present on 2nd index..  return the index..
```

```python
## packing of tuples..
pck = 2,3,4, 'harish', True
print(pck)
```

```python
## Unpaking of tuples..
first,second,a,b,c = pck
print(first, second)
```

```python
## Unpacking with *
num = (3,4,5,6,7)

first, *middle , last = num
print(first)
print(last)
print(middle)
```
