```python
## Dictionary...
## Empty dictionary..
empty_dict = {}
second_empty_dict = dict()

print(type(empty_dict))
print(type(second_empty_dict))
```

```python
student = {'name':"Krish", 'age':32, 'city':"Indore", "grade":"A"}
print(student)
print(type(student))
```

```python
## Accessing Dict element
student['age'] = 34
print(student)

nme =student.get('name')
print(nme)
```

```python
print(student.get("last_name" , "Not Available"))
```

```python
## NOTE:   Dictionary are mutable...
# you can add, delete, update the element of a dictionary..
student["last_name"] = "Jain"
print(student)
```

```python
# Delete...
del student["age"]
print(student)
```

```python
## Dictionary Methods...
# get all the keys
all_keys = student.keys()
print(all_keys)
all_values = student.values()
print(all_values)

itms = student.items()
print(itms)
```

```python
## Shallow Copy...
copy_student = student
copy_student['age'] = 43

print(student)
print(copy_student)
```

```python
st = student.copy()     # Shallow Copy
st['city_mp'] = "Bhopal"
print(st)
print(student)
```

```python
student['name'] = "Gopal"
print(st)
print(student)
```

```python
# Iterate over Dictionary...

print([k for k in student.keys()])
print([k for k in student.values()])
print([(k,v) for k,v in student.items()])
```
