```python
## Basic set operation...  

## Adding and Removing items..

mySet = {'amt', 3,4,5, 'gopal'}
print(mySet)
mySet.add("aaest")
print(mySet)
```

```python
mySet.remove('gopal')
print(mySet)
```

```python
mySet.remove("sdfsdf")  ## if key not found then will get error...
print(mySet)
```

```python
## remove element without getting error if item not found in it..

# use discard
mySet.discard('rajesh pat')
print(mySet)
```

```python
##  pop element will remove first element in set

set3 = {1,2,3,4,5,6,7}
set3.pop()
print(set3)
```

```python
set3.clear()
print(set3)
```

```python
# Mathematical Operations....
myset = {2,3,4,5,6}
myset2 = {6,7,8,9}

# Union
myset_union = myset.union(myset2)   ## UNION
myset_intersection = myset.intersection(myset2)  ## INTERSECTION
print(myset_intersection)
print(myset_union)
```

```python
M = myset.difference(myset2)  ## DIFFERENCE..
print(M)
```

```python
## Symetric Difference
a = {1,2,3,4,7,8,9}
b = {3,4,5,22,33,44}
print( a.symmetric_difference(b))    ## UNIQUE ELEMENT FROM BOTH THE SET...

```

```python
## SET METHODS.....

## issubset
## issuperset
## isdisjoint
myS = {1,2,3,4,5}
myS2 = {6,7,8,9}
print(myS.issubset(myS2))
print(myS.issuperset(myS2))
print(myS.isdisjoint(myS2))
```

```python
## FIND UNIQUE WORD COUNT...

myText = "This is my first example. This is okay but not first time."
lst = myText.split(" ")
unique_words = set(lst)
print(unique_words)
```
