<h1>How to use the Pandas .agg() function<h1>

```python
theme_by_year = myset.groupby('year').agg({'theme_id':pd.Series.nunique})
theme_by_year
```

```python
### Re-name  them_id column here

theme_by_year.rename(columns={'theme_id': "nr_themes"}, inplace=True)

print(theme_by_year.tail())
print(theme_by_year.head())
```

```python
import numpy as np

plt.ylim(0,1000)
# plt.yticks(np.arange(0,1000, step=100))
plt.plot(sets_by_year.index[:-2], sets_by_year.set_num[:-2])
plt.plot(theme_by_year.index[:-2], theme_by_year[:-2])
```

**Two Separate Axes**

```python

ax1 = plt.gca()
ax2 = ax1.twinx()


# plt.ylim(0,1000)
# plt.yticks(np.arange(0,1000, step=100))
ax1.plot(sets_by_year.index[:-2], sets_by_year.set_num[:-2], color='b')
ax2.plot(theme_by_year.index[:-2], theme_by_year[:-2], color='g')

ax1.set_xlabel("Years", fontsize=10)
ax1.set_ylabel("Number of Parts", fontsize=10, color='blue')
ax2.set_ylabel("Number of Themes", fontsize=10, color='green')

```
