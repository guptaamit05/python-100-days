```python
def calculate_love_score(name_one, name_two):
    ## "LOVE"  "YOU"
    combined_name = name_one+name_two
    print(combined_name)
    love_found = 0
    you_found = 0
    for x in combined_name:
        if x.lower() in "true":
            love_found+=1
        if x.lower() in "love":
            you_found+=1

    print(str(love_found)+str(you_found))

calculate_love_score("Kanye West", "Kim Kardashian")
    
```
