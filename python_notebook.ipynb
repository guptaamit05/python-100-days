```python
def calculate_love_score(name_one, name_two):
    ## "LOVE"  "YOU"
    combined_name = name_one+name_two
    print(combined_name)
    love_found = 0
    you_found = 0
    for x in combined_name:
        if x.lower() in "true":
            love_found+=1
        if x.lower() in "love":
            you_found+=1

    print(str(love_found)+str(you_found))

calculate_love_score("Kanye West", "Kim Kardashian")
    
```

```python
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def area(self):
        pass
    
class B(A):
    def __init__(self, radi):
        self.radi = radi
        self.__amount = 1000
    
    def area(self):
        return 3.14*(self.radi*self.radi)
    def access_amount(self):
        return self.__amount
b = B(4)
print(b.area())
# print(b.__amount)
print(b.access_amount())
```

```python
st = "abcabcd"
def find_longest_substring(st):
    max_len = 0
    left=0
    char_count = {}
    for x in range(len(st)):
        print(x)
        current = st[x]
        if(current in char_count.keys()  and char_count[current] >= left):
            left = char_count[current] +1
        max_len = max(max_len, x-left+1)
        
        char_count[current] = x
    return max_len

print(find_longest_substring(st))
```
